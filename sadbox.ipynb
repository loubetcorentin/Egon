```python
import torch
# TEXT TO SPEECH PARAMETERS
language = 'en'
model_id = 'v3_en'
sample_rate = 48000
speaker = 'en_1'
device = torch.device('cpu')

model_tts, example_text = torch.hub.load(repo_or_dir='snakers4/silero-models',
                                     model='silero_tts',
                                     language=language,
                                     speaker=model_id)                            
model_tts.to(device)  # gpu or cpu

model_tts.save_wav(text="This is boring as fuck",speaker=speaker,sample_rate=sample_rate)

```
