```python
import pyfirmata2
import time
import random
import threading
```

```python
PORT =  pyfirmata2.Arduino.AUTODETECT
```

```python
board = pyfirmata2.Arduino(PORT)
```

```python
pp = board.get_pin("d:9:s")
```

```python
class BufferServo():
    def __init__(self, pin):
        self.pin = pin
        self.current_value = 0
        self.goal_value = 0
        self.th = threading.Thread(target=self.threadloop)
        self.running=True
        self.th.start()
        
    def inch_closer(self):
        if self.goal_value> self.current_value:
            self.current_value +=1
        if self.goal_value< self.current_value:
            self.current_value -=1
        
    def threadloop(self):
        while self.running:
            self.inch_closer()
            self.send_msg()
            print(self.current_value)
            time.sleep(0.005)
        
    def send_msg(self):
        self.pin.write(self.current_value)
        
    def write(self,value):
        self.goal_value = value
        print(self.goal_value)
        
    def kill(self):
        self.running = False
    
bs = BufferServo(pp)
```
